## Writing core utilities and the main Streamlit interface into modular python scripts.

```python
#  Writing utils.py
%%writefile utils.py
import numpy as np
import pandas as pd
import shap
import matplotlib.pyplot as plt
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas
import io

def calculate_custom_features(amount, oldbalanceOrg, newbalanceOrig, oldbalanceDest, newbalanceDest):
    """
    Recalculates engineered features in real-time based on simulator input.
    """
    errorBalanceOrig = oldbalanceOrg - amount - newbalanceOrig
    errorBalanceDest = oldbalanceDest + amount - newbalanceDest
    
    return {
        'amount': amount,
        'oldbalanceOrg': oldbalanceOrg,
        'newbalanceOrig': newbalanceOrig,
        'oldbalanceDest': oldbalanceDest,
        'newbalanceDest': newbalanceDest,
        'errorBalanceOrig': errorBalanceOrig,
        'errorBalanceDest': errorBalanceDest
    }

def generate_llm_sar_narrative(transaction_data, risk_score, top_shap_features):
    """
    Simulates / Calls LLM to generate formal Suspicious Activity Report (SAR) narrative.
    """
    narrative = f"""
    SUSPICIOUS ACTIVITY REPORT (SAR) NARRATIVE
    --------------------------------------------------
    Automated Flagging Reason: High Fraud Probability ({risk_score*100:.1f}%)
    
    Incident Summary:
    Transaction involving account {transaction_data.get('step', 'N/A')} with an amount of ${transaction_data['amount']:,.2f} 
    was flagged by the XGBoost Anti-Fraud System.
    
    Key Primary Risk Drivers (SHAP Attribution):
    1. {top_shap_features[0]['feature']}: Value = {top_shap_features[0]['value']}, Impact = {top_shap_features[0]['shap_val']:.4f}
    2. {top_shap_features[1]['feature']}: Value = {top_shap_features[1]['value']}, Impact = {top_shap_features[1]['shap_val']:.4f}
    3. {top_shap_features[2]['feature']}: Value = {top_shap_features[2]['value']}, Impact = {top_shap_features[2]['shap_val']:.4f}
    
    Compliance Recommendation:
    Immediate hold placed on funds. Recommended for secondary human analyst review.
    --------------------------------------------------
    Report Generated automatically by PaySim XAI Engine.
    """
    return narrative.strip()

def create_pdf_report(narrative_text):
    """
    Generates downloadable PDF report from the text narrative.
    """
    buffer = io.BytesIO()
    p = canvas.Canvas(buffer, pagesize=letter)
    textobject = p.beginText()
    textobject.setTextOrigin(50, 750)
    textobject.setFont("Helvetica", 10)
    
    for line in narrative_text.split('\n'):
        textobject.textLine(line)
        
    p.drawText(textobject)
    p.showPage()
    p.save()
    buffer.seek(0)
    return buffer
```
